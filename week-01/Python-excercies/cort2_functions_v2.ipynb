## APPROACH_2: Convert the above to Functions - to modularize

```python
## 1. FUNCTION
def calculate_receipt(item_prices_dict):
    receipt_dict = {}
    total_cost = 0
    
    # 2. Calculate  -- i.Get quantity ii. calculate cost for each tiem
    # For each of the 'fruit' i.e. item:
    for item in item_prices_dict:
        item_qty = int(input(f"How many {item} do you need :"))
        item_price = item_prices_dict[item] # Get the price of 'apple'
        cost = item_qty * item_price
    
        # Capture what we want as part of the RECEIPT - DOWNSTREAM
        receipt_dict[item] = {
            "price" : item_price,
            "quantity" : item_qty,
            "cost" : cost,
        }
        total_cost = total_cost + cost
        # print(f"-->total_cost:{total_cost};  receipt_dict: {receipt_dict} ")

    # RETURN the processed info
    return total_cost, receipt_dict
```

```python
## 2. Function
def generate_receipt(receipt_dict):
    ##### OUTSIDE the for loop - calculate the i. balance ii. Generate the RECEIPT
    balance = bill_paid_by_customer - total_cost
    
    ##### Generate Receipt #####
    print("""########## REIPT ##########""")
    print("""#############################""")
    for item in receipt_dict:
        item_details = receipt_dict[item]
        print(f"--> {item} @ {item_details['price']} * {item_details['quantity']} : {item_details['cost']}")
        # 'price': 2.25, 'quantity': 2, 'cost': 4.5
    
    print("""#############################""")
    print(f"-->  Total Balance: {balance} ")
    print("""#############################""")
```

```python
## 2. MAIN/DRIVER code

## 1. Input:
# Customer selected the below item/fruit
# item_prices_dict = {
#     "apple" : 3.5,
#     "orange" : 2.25,
#     "mango" : 4.0,
#     "banana" : 1.75,
#     "grapes" : 4.75
# }
item_prices_dict = {
    "apple" : 3.5,
    "orange" : 2.25
}

bill_paid_by_customer = 100

##### Call the function to calculate and generate a report
'''
input:
    item_prices_dict: dict
return: 
        total_cost, receipt_dict
'''
print(f"item_prices_dict:{item_prices_dict}")

# Task1: Calculations/Transformations
total_cost, receipt_dict = calculate_receipt(item_prices_dict)

# Task2: Build a Report/REceipt
generate_receipt(receipt_dict)
```
