## Q1 (Loops): Write a loop that prints multiples of 5 from 5 to 100.


```python
# A classical method where a for loop has been used with a condition providing the range function.Here the "end'='" makes the
#output visible in a line
for i in range(5,101):
    if i%5==0:
        print(i,end=' ' )
    
```

```python
# same problem solved by using the skip functionality
for i in range(5,101,5):
    print(i,end = ' ')
```

```python
#THIS has been reduced to 1 liner with a variable assigned,.
l =[i for i in range(5,101,5)]
```

```python
print(l)
```

#### we can make it more simple by using the list comprehension style

```python
y = [i for i in range(5,101) if i%5==0]
print(y,end='')
```

### Q2 (range): Use range() to generate a list of numbers from 50 down to 0.



```python
for i in range(50,-1,-1):
    print(i,end = ' ')
```

### Q3 (List comprehension): Use list comprehension to filter out numbers divisible by 3 from a list of 1–30.


```python
l = list(range(1,31))
f = [i for i in l if i%3==0]
print(f)
```

### Q4 (Strings): Write code to print each character of a string on a new line.

```python
# In this method we are listing each digit of the python in a line  
text = 'python'
for i in list(text):
    print(i,end= '\n')
```

```python
### This approach is by using enumerate function where i,y two variables are used as enumerate functions where the i is index value and y is the individual letter in the list
```

```python
for i,y in enumerate(list(text)):
    print(i,y)
```

### Q5 (CSV): Write Python code to write a list of numbers to a JSON file.Sample input: nums = [10, 20, 30, 40]



```python
import json

nums = [10, 20, 30, 40]

with open("numbers.json","w") as json_file:
    json.dump(nums, json_file)

print("--> Completed writing to the JSON file the numbers...")
```
